```python
# 복습 - 1부터 10까지의 자연수 중에서 홀수와 홀수들의 합
# 아래 출력예시
# 1, 3, 5, 7, 9
# 1~10까지의 홀수의 합: 25

hap = 0
for i in range(1, 11):
    if i % 2 == 1:
        print(i, end=', ')
        hap += i
# print()
print(f'\n1~10까지의 홀수의 합: {hap}')
```

```python
hap = 0
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=', ')
    hap += i

print(f'\n1~10까지의 홀수의 합: {hap}')
```

```python
# 20보다 작은 정수 중에서 3의 배수와 5의 배수를 제외하고 출력
x = 0

while True:
    x += 1
    if x > 20:
        break
    if (x % 3 == 0) or (x % 5 == 0):
        continue
    print(x, end=' ')
    
# 1 2 4  ....
```

```python
# 3, 6, 9로 끝나는 숫자에 '짝'을 출력하는 369 게임을 작성해주세요.
# 1 2 짝 4 5 짝 7 8 짝 10 11 12 짝 14 15 짝 17 18 짝 20

# num = int(input('1부터 얼마까지 진행할까요?(예:20): '))
num = 20

for i in range(1, num+1):
    if (i % 10 == 3) or (i % 10 == 6) or (i % 10 == 9):
        print('짝', end=' ')
    else:
        print(i, end=' ')
```
